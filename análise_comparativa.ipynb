# Análise comparativa de representações
Execute as células para gerar a tabela pedida. Embeddings exigem o download inicial do modelo.

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from ouvidoria_core import load_manifestacoes, embed_texts, MODEL_OPTIONS
base=load_manifestacoes(); textos=[x['texto'] for x in base]
ids=[x['id'] for x in base]; pares=[('M003','M017'),('M008','M022'),('M008','M031')]
```

```python
def score(matriz,a,b):
    i,j=ids.index(a),ids.index(b); return cosine_similarity(matriz[i:i+1],matriz[j:j+1])[0,0]
bow=CountVectorizer(stop_words=None).fit_transform(textos)
tfidf=TfidfVectorizer().fit_transform(textos)
emb=embed_texts(textos, MODEL_OPTIONS['Multilíngue MiniLM (recomendado)'])
pd.DataFrame([{'par':f'{a} × {b}','BoW':score(bow,a,b),'TF-IDF':score(tfidf,a,b),'Embedding':score(emb,a,b)} for a,b in pares]).round(3)
```

**Conclusão.** BoW recompensa termos repetidos; TF-IDF reduz o efeito de palavras frequentes, mas ambos dependem de vocabulário comum. Embeddings tendem a reconhecer paráfrases como *buraco* e *asfalto esburacado*, mantendo baixa similaridade entre saúde e iluminação.
